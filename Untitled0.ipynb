```python
# ============================================================
# INTELLIGENT EXAM ANALYTICS DASHBOARD  (fixed + colourful version)
# Google Colab + Streamlit + Plotly  -- paste into ONE Colab cell
#
# FIX FOR "TypeError: Failed to fetch dynamically imported module ...":
#   Streamlit loads its widgets (Multiselect, Metric, ...) as separate
#   JavaScript files. The free localtunnel service often blocks or drops
#   these requests, so every widget showed that error. The app code was
#   not the problem. This version uses a Cloudflare tunnel first (stable),
#   keeps localtunnel only as a backup, and turns off the Streamlit
#   settings that break behind tunnels.
# ============================================================

# ------------------------------------------------------------
# 1. INSTALL REQUIRED LIBRARIES
# ------------------------------------------------------------
!pip -q install streamlit plotly pandas numpy
!npm install -g localtunnel -q

# ------------------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------------------
import os, sys, re, glob, time, subprocess, urllib.request
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 3. FIND / UPLOAD DATASET  (falls back to a DEMO dataset)
# ------------------------------------------------------------
DEFAULT_PATH = "/content/intelligent_exam_5k_high_accuracy_dataset.csv"
DATA_PATH = None

def make_demo_dataset(path, n=5000, seed=42):
    """Synthetic question-level data, used ONLY when no CSV is available."""
    rng = np.random.default_rng(seed)
    subjects = ["Mathematics", "Physics", "Chemistry", "Biology", "Computer Science"]
    blooms   = ["Remember", "Understand", "Apply", "Analyze", "Evaluate", "Create"]
    diffs    = ["easy", "medium", "hard"]
    diff     = rng.choice(diffs, n, p=[0.35, 0.40, 0.25])
    base     = pd.Series(diff).map({"easy": 0.78, "medium": 0.58, "hard": 0.38}).values
    correct  = np.clip(base + rng.normal(0, 0.10, n), 0.05, 0.98)
    attempts = rng.integers(80, 400, n)
    corr_n   = (correct * attempts).round().astype(int)
    words    = rng.integers(8, 60, n)
    demo = pd.DataFrame({
        "question_text": [f"Demo question {i+1}: explain concept {i%97}" for i in range(n)],
        "subject": rng.choice(subjects, n),
        "cognitive_level_bloom": rng.choice(blooms, n, p=[.2, .2, .25, .15, .12, .08]),
        "difficulty_label": diff,
        "readability_score": np.clip(rng.normal(60, 12, n), 10, 100).round(1),
        "word_count": words,
        "sentence_count": np.maximum(1, (words / rng.integers(8, 20, n)).round()).astype(int),
        "time_taken_minutes": np.clip(rng.normal(4, 1.5, n) + (1 - correct) * 3, 0.5, 15).round(2),
        "total_students_attempted": attempts,
        "correct_attempts": corr_n,
        "incorrect_attempts": attempts - corr_n,
        "correct_percentage": correct.round(3),
        "learning_gap_score": np.clip(1 - correct + rng.normal(0, 0.05, n), 0, 1).round(3),
        "discrimination_index": np.clip(rng.normal(0.35, 0.15, n), -0.1, 0.9).round(3),
        "assessment_quality_score": np.clip(rng.normal(0.70, 0.12, n), 0.1, 1).round(3),
    })
    demo.to_csv(path, index=False)

if os.path.exists(DEFAULT_PATH):
    DATA_PATH = DEFAULT_PATH
else:
    found = sorted(glob.glob("/content/*.csv"))
    preferred = [f for f in found if "exam" in os.path.basename(f).lower()]
    if preferred or found:
        DATA_PATH = (preferred or found)[0]
    else:
        print("Dataset not found at:", DEFAULT_PATH)
        print("Please upload your CSV file (cancel to use a DEMO dataset).")
        try:
            from google.colab import files
            uploaded = files.upload()
            csvs = [f for f in uploaded if f.lower().endswith(".csv")]
            if csvs:
                DATA_PATH = "/content/" + csvs[0]
        except Exception as e:
            print("Upload not available:", e)

if DATA_PATH is None:
    DATA_PATH = "/content/demo_exam_dataset.csv"
    make_demo_dataset(DATA_PATH)
    print("\n*** WARNING: no CSV supplied - using a synthetic DEMO dataset. ***")

print("\nDataset path:", DATA_PATH)

# ------------------------------------------------------------
# 4. TEST DATASET
# ------------------------------------------------------------
try:
    test_df = pd.read_csv(DATA_PATH)
except Exception as e:
    raise RuntimeError(f"Unable to read dataset: {e}")

print("\n" + "=" * 60)
print("DATASET LOADED SUCCESSFULLY")
print("=" * 60)
print("Rows   :", test_df.shape[0])
print("Columns:", test_df.shape[1])
print(test_df.columns.tolist())

# ------------------------------------------------------------
# 5. STREAMLIT APPLICATION (written to a file)
# ------------------------------------------------------------
APP_CODE = r'''
import os
import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

st.set_page_config(page_title="Intelligent Exam Analytics", page_icon="🧠",
                   layout="wide", initial_sidebar_state="expanded")

# ---------- colourful KPI cards ----------
st.markdown("""
<style>
div[data-testid="stMetric"]{
  background:linear-gradient(135deg,rgba(99,102,241,.35),rgba(236,72,153,.25));
  border:1px solid rgba(255,255,255,.12);border-radius:14px;padding:14px 16px;
  box-shadow:0 4px 14px rgba(0,0,0,.25);}
div[data-testid="stMetricValue"]{font-weight:700;}
</style>
""", unsafe_allow_html=True)

# ---------- version-safe wrappers (use_container_width was removed) ----------
def _ver(v):
    out = []
    for p in v.split(".")[:3]:
        d = "".join(ch for ch in p if ch.isdigit())
        out.append(int(d) if d else 0)
    return tuple(out + [0] * (3 - len(out)))

NEW_API = _ver(st.__version__) >= (1, 50, 0)

def _try(fn, obj, **extra):
    options = [{"width": "stretch"}, {"use_container_width": True}, {}]
    if not NEW_API:
        options = [{"use_container_width": True}, {}]
    last = None
    for kw in options:
        try:
            return fn(obj, **kw, **extra)
        except Exception as e:
            last = e
    raise last

def show_chart(fig):
    fig.update_layout(margin=dict(l=20, r=20, t=60, b=20))
    _try(st.plotly_chart, fig)

def show_table(d, **extra):
    _try(st.dataframe, d, **extra)

# ---------- colour palettes ----------
VIVID = px.colors.qualitative.Vivid
BOLD = px.colors.qualitative.Bold
DIFF_COLORS = {"Easy": "#22c55e", "Medium": "#f59e0b", "Hard": "#ef4444"}
GOOD_TO_BAD = ["#22c55e", "#84cc16", "#facc15", "#f97316", "#ef4444"]   # gap bands
BAD_TO_GOOD = GOOD_TO_BAD[::-1]                                         # performance/quality

def colors_for(labels, palette):
    return {l: palette[i % len(palette)] for i, l in enumerate(labels)}

# ---------- load data ----------
DATA_PATH = os.environ.get("EXAM_DATA_PATH", "")

@st.cache_data(show_spinner=False)
def load_data(path):
    return pd.read_csv(path)

try:
    df = load_data(DATA_PATH).copy()
except Exception as e:
    st.error(f"Error loading dataset '{DATA_PATH}': {e}")
    st.stop()

# ---------- normalise column names ----------
def norm(c):
    return str(c).strip().lower().replace(" ", "_").replace("-", "_")

df.columns = [norm(c) for c in df.columns]

ALIASES = {
    "question": "question_text", "questions": "question_text",
    "bloom_level": "cognitive_level_bloom", "bloom": "cognitive_level_bloom",
    "cognitive_level": "cognitive_level_bloom",
    "difficulty": "difficulty_label", "difficulty_level": "difficulty_label",
    "time_taken": "time_taken_minutes", "time_minutes": "time_taken_minutes",
    "students_attempted": "total_students_attempted",
    "correct_pct": "correct_percentage", "learning_gap": "learning_gap_score",
    "quality_score": "assessment_quality_score",
}
df = df.rename(columns={k: v for k, v in ALIASES.items()
                        if k in df.columns and v not in df.columns})

CORE = ["question_text", "subject", "cognitive_level_bloom", "difficulty_label",
        "correct_percentage", "learning_gap_score", "discrimination_index",
        "assessment_quality_score"]
OPTIONAL = ["readability_score", "word_count", "sentence_count", "time_taken_minutes",
            "total_students_attempted", "correct_attempts", "incorrect_attempts"]

missing = [c for c in CORE if c not in df.columns]
if missing:
    st.error("Required columns missing from the dataset: " + ", ".join(missing))
    st.write("Columns found in your file:", list(df.columns))
    st.stop()
for c in OPTIONAL:
    if c not in df.columns:
        df[c] = np.nan

# ---------- numeric conversion ----------
NUMERIC = CORE[4:] + OPTIONAL
for c in NUMERIC:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df = df.dropna(subset=CORE[4:]).copy()
if df.empty:
    st.error("No valid numeric rows remain after cleaning. Check the dataset values.")
    st.stop()

# values stored as 0-100 are converted to 0-1
for c in ["correct_percentage", "learning_gap_score",
          "assessment_quality_score", "discrimination_index"]:
    if df[c].abs().max() > 1.5:
        df[c] = df[c] / 100.0

# ---------- text cleaning ----------
def clean_text(s):
    s = s.fillna("Unknown").astype(str).str.strip()
    return s.replace({"": "Unknown", "nan": "Unknown", "None": "Unknown"})

df["question_text"] = clean_text(df["question_text"])
df["subject"] = clean_text(df["subject"])
df["cognitive_level_bloom"] = clean_text(df["cognitive_level_bloom"]).str.title()

DIFF_MAP = {"easy": "Easy", "simple": "Easy", "medium": "Medium",
            "moderate": "Medium", "hard": "Hard", "difficult": "Hard"}
d_raw = clean_text(df["difficulty_label"])
df["difficulty_label"] = d_raw.str.lower().map(DIFF_MAP).fillna(d_raw.str.title())

difficulty_order = ["Easy", "Medium", "Hard"]
difficulty_order += sorted(set(df["difficulty_label"]) - set(difficulty_order))
BLOOM_STD = ["Remember", "Understand", "Apply", "Analyze", "Evaluate", "Create"]
present = set(df["cognitive_level_bloom"])
bloom_order = [b for b in BLOOM_STD if b in present] + sorted(present - set(BLOOM_STD))

subjects_all = sorted(df["subject"].unique())
SUBJ_COLORS = colors_for(subjects_all, BOLD)
BLOOM_COLORS = colors_for(bloom_order, VIVID)
DIFF_COLORS = {**colors_for(difficulty_order, VIVID), **DIFF_COLORS}

# ---------- derived fields ----------
df["correct_percentage_pct"] = df["correct_percentage"] * 100
df["learning_gap_pct"] = df["learning_gap_score"] * 100
df["discrimination_pct"] = df["discrimination_index"] * 100
df["quality_pct"] = df["assessment_quality_score"] * 100
df["attempts_size"] = df["total_students_attempted"].fillna(1).clip(lower=1)

def band(series, edges, labels):
    return pd.cut(series, bins=[-np.inf] + edges + [np.inf],
                  labels=labels, right=False).astype(object)

performance_order = ["Very Low", "Low", "Moderate", "High", "Very High"]
gap_order = ["Minimal Gap", "Low Gap", "Moderate Gap", "High Gap", "Critical Gap"]
# discrimination bands are spread over 0.6-1.0 so that a dataset whose values
# lie between 0.5 and 1.0 does not collapse into one single bar
disc_order = ["Below 0.60", "0.60 - 0.70", "0.70 - 0.80", "0.80 - 0.90", "0.90 and above"]
quality_order = ["Poor", "Needs Improvement", "Good", "Very Good", "Excellent"]

df["performance_band"] = band(df["correct_percentage"], [0.30, 0.50, 0.70, 0.85], performance_order)
df["learning_gap_band"] = band(df["learning_gap_score"], [0.20, 0.40, 0.60, 0.80], gap_order)
df["discrimination_band"] = band(df["discrimination_index"], [0.60, 0.70, 0.80, 0.90], disc_order)
df["quality_band"] = band(df["assessment_quality_score"], [0.40, 0.60, 0.75, 0.90], quality_order)

PERF_COLORS = dict(zip(performance_order, BAD_TO_GOOD))
GAP_COLORS = dict(zip(gap_order, GOOD_TO_BAD))
DISC_COLORS = dict(zip(disc_order, ["#6366f1", "#06b6d4", "#10b981", "#f59e0b", "#ec4899"]))
QUAL_COLORS = dict(zip(quality_order, BAD_TO_GOOD))

# ---------- sidebar ----------
st.sidebar.title("🎛️ Controls")
st.sidebar.markdown("### Data Filters")
selected_subjects = st.sidebar.multiselect("Select Subject", subjects_all, default=subjects_all)
selected_bloom = st.sidebar.multiselect("Bloom's Cognitive Level", bloom_order, default=bloom_order)
selected_difficulty = st.sidebar.multiselect("Difficulty", difficulty_order, default=difficulty_order)

filtered_df = df[
    df["subject"].isin(selected_subjects)
    & df["cognitive_level_bloom"].isin(selected_bloom)
    & df["difficulty_label"].isin(selected_difficulty)
].copy()

st.sidebar.markdown("---")
st.sidebar.info("Use these filters to dynamically adjust metrics across all tabs.")

# ---------- header ----------
st.title("🧠 Intelligent Exam Analytics Dashboard")
st.markdown("**Question-Level Assessment Intelligence • Learning Analytics • Quality Analysis**")
st.caption(f"Showing {len(filtered_df):,} questions out of {len(df):,} total records.")

if filtered_df.empty:
    st.warning("No records match the selected filters. Please adjust the filters.")
    st.stop()

# ---------- KPIs ----------
c1, c2, c3, c4, c5 = st.columns(5)
c1.metric("📚 Questions", f"{len(filtered_df):,}")
c2.metric("🎯 Avg Correct", f"{filtered_df['correct_percentage'].mean() * 100:.1f}%")
c3.metric("🔍 Avg Learning Gap", f"{filtered_df['learning_gap_score'].mean() * 100:.1f}%")
c4.metric("📈 Avg Discrimination", f"{filtered_df['discrimination_index'].mean():.2f}")
c5.metric("⭐ Avg Quality", f"{filtered_df['assessment_quality_score'].mean() * 100:.1f}%")

tabs = st.tabs([
    "🏠 Overview", "📚 Subject Analysis", "🧠 Bloom Analysis", "🎯 Difficulty",
    "👨‍🎓 Performance", "🔍 Learning Gap", "📈 Discrimination",
    "📝 Assessment Quality", "🔗 Relationships", "📋 Question Explorer",
])

def counts_by(series, order, name):
    out = series.value_counts().reindex(order).fillna(0).astype(int).reset_index()
    out.columns = [name, "Questions"]
    return out

def cbar(data, x, y, cmap, title, fmt=None, order=None, ytitle=None):
    """Bar chart where every bar has its own colour."""
    kw = {}
    if order:
        kw["category_orders"] = {x: order}
    fig = px.bar(data, x=x, y=y, text=y, color=x, color_discrete_map=cmap,
                 title=title, **kw)
    fig.update_traces(texttemplate=fmt or "%{text}", textposition="outside",
                      marker_line_width=0, cliponaxis=False)
    fig.update_layout(showlegend=False)
    if ytitle:
        fig.update_layout(yaxis_title=ytitle)
    return fig

# ---------- TAB 1 ----------
with tabs[0]:
    st.subheader("📊 Examination Overview")
    a, b = st.columns(2)
    with a:
        sc = filtered_df["subject"].value_counts().reset_index()
        sc.columns = ["subject", "count"]
        fig = px.treemap(sc, path=["subject"], values="count", color="subject",
                         color_discrete_map=SUBJ_COLORS,
                         title="Question Distribution by Subject")
        fig.update_traces(textinfo="label+value+percent root")
        fig.update_layout(height=450)
        show_chart(fig)
    with b:
        dc = counts_by(filtered_df["difficulty_label"].astype(str), difficulty_order, "difficulty")
        fig = cbar(dc, "difficulty", "Questions", DIFF_COLORS, "Difficulty Distribution",
                   order=difficulty_order)
        fig.update_layout(height=450)
        show_chart(fig)
    st.markdown("### 🧠 Cognitive Distribution")
    bs = (filtered_df.groupby(["subject", "cognitive_level_bloom"])
          .size().reset_index(name="count"))
    fig = px.sunburst(bs, path=["subject", "cognitive_level_bloom"], values="count",
                      color="subject", color_discrete_map=SUBJ_COLORS,
                      title="Subject → Bloom's Cognitive Level")
    fig.update_layout(height=600)
    show_chart(fig)

# ---------- TAB 2 ----------
with tabs[1]:
    st.subheader("📚 Subject-Wise Analysis")
    ss = (filtered_df.groupby("subject").agg(
        Questions=("question_text", "count"),
        Avg_Correct=("correct_percentage", "mean"),
        Avg_Learning_Gap=("learning_gap_score", "mean"),
        Avg_Discrimination=("discrimination_index", "mean"),
        Avg_Quality=("assessment_quality_score", "mean"),
        Avg_Time=("time_taken_minutes", "mean"),
    ).reset_index())
    ss["Avg_Correct"] *= 100
    ss["Avg_Learning_Gap"] *= 100
    ss["Avg_Quality"] *= 100

    a, b = st.columns(2)
    with a:
        fig = cbar(ss, "subject", "Avg_Correct", SUBJ_COLORS,
                   "Average Correct Percentage by Subject", "%{text:.1f}%",
                   ytitle="Correct Percentage (%)")
        show_chart(fig)
    with b:
        fig = cbar(ss, "subject", "Avg_Learning_Gap", SUBJ_COLORS,
                   "Learning Gap by Subject", "%{text:.1f}%", ytitle="Learning Gap (%)")
        show_chart(fig)

    st.markdown("### 🕸️ Subject Performance Profile")
    labels = ["Correct %", "Learning Retention", "Discrimination", "Quality"]
    fig = go.Figure()
    for _, row in ss.iterrows():
        vals = [row["Avg_Correct"], 100 - row["Avg_Learning_Gap"],
                row["Avg_Discrimination"] * 100, row["Avg_Quality"]]
        col = SUBJ_COLORS.get(row["subject"], "#6366f1")
        fig.add_trace(go.Scatterpolar(r=vals + [vals[0]], theta=labels + [labels[0]],
                                      fill="toself", name=row["subject"],
                                      line=dict(color=col), opacity=0.55))
    fig.update_layout(polar=dict(radialaxis=dict(visible=True, range=[0, 100])), height=550)
    show_chart(fig)
    st.subheader("📋 Subject Summary Table")
    show_table(ss.round(3))

# ---------- TAB 3 ----------
with tabs[2]:
    st.subheader("🧠 Bloom's Cognitive Level Analysis")
    bsum = (filtered_df.groupby("cognitive_level_bloom").agg(
        Questions=("question_text", "count"),
        Avg_Correct=("correct_percentage", "mean"),
        Avg_Gap=("learning_gap_score", "mean"),
        Avg_Time=("time_taken_minutes", "mean"),
        Avg_Quality=("assessment_quality_score", "mean"),
    ).reset_index())
    bsum["Avg_Correct"] *= 100
    bsum["Avg_Gap"] *= 100
    bsum["Avg_Quality"] *= 100
    bsum["_o"] = bsum["cognitive_level_bloom"].map({b: i for i, b in enumerate(bloom_order)})
    bsum = bsum.sort_values("_o").drop(columns="_o")

    a, b = st.columns(2)
    with a:
        fig = cbar(bsum, "cognitive_level_bloom", "Avg_Correct", BLOOM_COLORS,
                   "Performance Across Bloom's Levels", "%{text:.1f}%", order=bloom_order,
                   ytitle="Average Correct (%)")
        show_chart(fig)
    with b:
        fig = px.bar(bsum, x="cognitive_level_bloom", y="Avg_Gap", text="Avg_Gap",
                     color="Avg_Gap", color_continuous_scale="Turbo",
                     category_orders={"cognitive_level_bloom": bloom_order},
                     title="Average Learning Gap by Bloom's Level")
        fig.update_traces(texttemplate="%{text:.1f}%", textposition="outside", cliponaxis=False)
        fig.update_layout(yaxis_title="Learning Gap (%)", coloraxis_showscale=False)
        show_chart(fig)

    a, b = st.columns(2)
    with a:
        fig = px.box(filtered_df, x="cognitive_level_bloom", y="correct_percentage_pct",
                     color="cognitive_level_bloom", color_discrete_map=BLOOM_COLORS,
                     category_orders={"cognitive_level_bloom": bloom_order},
                     title="Correct Percentage Spread by Bloom's Level")
        fig.update_layout(showlegend=False, yaxis_title="Correct Percentage (%)")
        show_chart(fig)
    with b:
        if bsum["Avg_Time"].notna().any():
            fig = px.scatter(bsum, x="Avg_Time", y="Avg_Correct", size="Questions",
                             color="cognitive_level_bloom", color_discrete_map=BLOOM_COLORS,
                             hover_name="cognitive_level_bloom", size_max=45,
                             title="Cognitive Complexity vs Performance")
            fig.update_layout(xaxis_title="Average Time (minutes)",
                              yaxis_title="Average Correct (%)")
            show_chart(fig)
        else:
            st.info("time_taken_minutes is not available, so this chart is skipped.")
    show_table(bsum.round(3))

# ---------- TAB 4 ----------
with tabs[3]:
    st.subheader("🎯 Difficulty Analysis")
    dsum = (filtered_df.groupby("difficulty_label", observed=True).agg(
        Questions=("question_text", "count"),
        Avg_Correct=("correct_percentage", "mean"),
        Avg_Time=("time_taken_minutes", "mean"),
        Avg_Gap=("learning_gap_score", "mean"),
        Avg_Discrimination=("discrimination_index", "mean"),
        Avg_Quality=("assessment_quality_score", "mean"),
    ).reset_index())
    dsum["Avg_Correct"] *= 100
    dsum["Avg_Gap"] *= 100
    dsum["Avg_Quality"] *= 100
    dsum["_o"] = dsum["difficulty_label"].map({d: i for i, d in enumerate(difficulty_order)})
    dsum = dsum.sort_values("_o").drop(columns="_o")

    a, b = st.columns(2)
    with a:
        fig = cbar(dsum, "difficulty_label", "Avg_Correct", DIFF_COLORS,
                   "Correct Percentage by Difficulty", "%{text:.1f}%", order=difficulty_order,
                   ytitle="Correct Percentage (%)")
        show_chart(fig)
    with b:
        fig = px.box(filtered_df, x="difficulty_label", y="correct_percentage_pct",
                     color="difficulty_label", color_discrete_map=DIFF_COLORS,
                     category_orders={"difficulty_label": difficulty_order},
                     title="Correct Percentage Spread by Difficulty")
        fig.update_layout(showlegend=False, yaxis_title="Correct Percentage (%)")
        show_chart(fig)

    if filtered_df["time_taken_minutes"].notna().any():
        sample = filtered_df if len(filtered_df) <= 3000 else filtered_df.sample(3000, random_state=1)
        fig = px.scatter(sample, x="time_taken_minutes", y="correct_percentage_pct",
                         color="difficulty_label", color_discrete_map=DIFF_COLORS,
                         size="attempts_size", size_max=12, opacity=0.6,
                         category_orders={"difficulty_label": difficulty_order},
                         hover_data=["subject", "cognitive_level_bloom"],
                         title="Time vs Performance by Difficulty")
        fig.update_layout(xaxis_title="Time Taken (minutes)",
                          yaxis_title="Correct Percentage (%)")
        show_chart(fig)
    else:
        st.info("time_taken_minutes is not available, so this chart is skipped.")
    st.markdown("### Subject vs Difficulty")
    sd = filtered_df.groupby(["subject", "difficulty_label"]).size().reset_index(name="Questions")
    fig = px.bar(sd, x="subject", y="Questions", color="difficulty_label",
                 color_discrete_map=DIFF_COLORS, barmode="stack", text="Questions",
                 category_orders={"difficulty_label": difficulty_order},
                 title="Questions per Subject, Split by Difficulty")
    show_chart(fig)
    show_table(dsum.round(3))

# ---------- TAB 5 ----------
with tabs[4]:
    st.subheader("👨‍🎓 Performance Analysis")
    st.info("The dataset contains question-level assessment records rather than student IDs.")
    pc = counts_by(filtered_df["performance_band"], performance_order, "Performance Band")
    a, b = st.columns(2)
    with a:
        fig = cbar(pc, "Performance Band", "Questions", PERF_COLORS,
                   "Performance Band Distribution", order=performance_order)
        show_chart(fig)
    with b:
        fig = px.pie(pc, names="Performance Band", values="Questions", hole=0.55,
                     color="Performance Band", color_discrete_map=PERF_COLORS,
                     category_orders={"Performance Band": performance_order},
                     title="Overall Performance Split")
        show_chart(fig)
    if filtered_df["total_students_attempted"].notna().any():
        tot = float(filtered_df["total_students_attempted"].sum())
        cor = float(filtered_df["correct_attempts"].sum())
        inc = float(filtered_df["incorrect_attempts"].sum())
        if tot > 0 and (cor + inc) > 0:
            ci = pd.DataFrame({"Result": ["Correct attempts", "Incorrect attempts"],
                               "Attempts": [cor, inc]})
            fig = px.pie(ci, names="Result", values="Attempts", hole=0.5, color="Result",
                         color_discrete_map={"Correct attempts": "#22c55e",
                                             "Incorrect attempts": "#ef4444"},
                         title=f"Correct vs Incorrect Attempts (total {int(tot):,})")
            show_chart(fig)
    fig = px.histogram(filtered_df, x="correct_percentage_pct", nbins=30,
                       color="difficulty_label", color_discrete_map=DIFF_COLORS,
                       category_orders={"difficulty_label": difficulty_order},
                       title="Correct Percentage Distribution (coloured by difficulty)")
    fig.update_layout(xaxis_title="Correct Percentage (%)", yaxis_title="Questions")
    show_chart(fig)

# ---------- TAB 6 ----------
with tabs[5]:
    st.subheader("🔍 Learning Gap Analysis")
    gs = counts_by(filtered_df["learning_gap_band"], gap_order, "Learning Gap")
    a, b = st.columns(2)
    with a:
        fig = cbar(gs, "Learning Gap", "Questions", GAP_COLORS,
                   "Learning Gap Distribution", order=gap_order)
        show_chart(fig)
    with b:
        sg = (filtered_df.groupby("subject")["learning_gap_score"].mean()
              .mul(100).reset_index(name="Learning Gap %"))
        fig = cbar(sg, "subject", "Learning Gap %", SUBJ_COLORS,
                   "Average Learning Gap by Subject", "%{text:.1f}%")
        show_chart(fig)
    fig = px.histogram(filtered_df, x="learning_gap_pct", nbins=30, marginal="box",
                       color="difficulty_label", color_discrete_map=DIFF_COLORS,
                       category_orders={"difficulty_label": difficulty_order},
                       title="Learning Gap Score Distribution (coloured by difficulty)")
    fig.update_layout(xaxis_title="Learning Gap (%)", yaxis_title="Questions")
    show_chart(fig)

# ---------- TAB 7 ----------
with tabs[6]:
    st.subheader("📈 Discrimination Index Analysis")
    ds = counts_by(filtered_df["discrimination_band"], disc_order, "Discrimination Level")
    a, b = st.columns(2)
    with a:
        fig = cbar(ds, "Discrimination Level", "Questions", DISC_COLORS,
                   "Discrimination Index Classification", order=disc_order)
        show_chart(fig)
    with b:
        fig = px.histogram(filtered_df, x="discrimination_index", nbins=25, marginal="box",
                           color="difficulty_label", color_discrete_map=DIFF_COLORS,
                           category_orders={"difficulty_label": difficulty_order},
                           title="Distribution of Discrimination Index")
        show_chart(fig)
    fig = px.box(filtered_df, x="cognitive_level_bloom", y="discrimination_index",
                 color="cognitive_level_bloom", color_discrete_map=BLOOM_COLORS,
                 category_orders={"cognitive_level_bloom": bloom_order},
                 title="Discrimination Index by Bloom's Level")
    fig.update_layout(showlegend=False)
    show_chart(fig)

# ---------- TAB 8 ----------
with tabs[7]:
    st.subheader("📝 Assessment Quality Analysis")
    qs = counts_by(filtered_df["quality_band"], quality_order, "Quality Band")
    a, b = st.columns(2)
    with a:
        fig = px.pie(qs, names="Quality Band", values="Questions", hole=0.58,
                     color="Quality Band", color_discrete_map=QUAL_COLORS,
                     category_orders={"Quality Band": quality_order},
                     title="Assessment Quality Distribution")
        show_chart(fig)
    with b:
        fig = px.histogram(filtered_df, x="quality_pct", nbins=25, marginal="box",
                           color="difficulty_label", color_discrete_map=DIFF_COLORS,
                           category_orders={"difficulty_label": difficulty_order},
                           title="Assessment Quality Score Distribution")
        fig.update_layout(xaxis_title="Quality Score (%)", yaxis_title="Questions")
        show_chart(fig)
    qd = (filtered_df.groupby("cognitive_level_bloom")["quality_pct"].mean()
          .reset_index(name="Avg Quality %"))
    fig = cbar(qd, "cognitive_level_bloom", "Avg Quality %", BLOOM_COLORS,
               "Average Assessment Quality by Bloom's Level", "%{text:.1f}%", order=bloom_order)
    show_chart(fig)

# ---------- TAB 9 ----------
with tabs[8]:
    st.subheader("🔗 Relationship Between Variables")
    rel = ["readability_score", "word_count", "sentence_count", "time_taken_minutes",
           "total_students_attempted", "correct_percentage", "learning_gap_score",
           "discrimination_index", "assessment_quality_score"]
    rel = [c for c in rel if filtered_df[c].notna().sum() > 1 and filtered_df[c].nunique() > 1]
    if len(rel) >= 2:
        corr = filtered_df[rel].corr()
        fig = px.imshow(corr, text_auto=".2f", aspect="auto",
                        color_continuous_scale="Spectral", zmin=-1, zmax=1,
                        title="Correlation Matrix of Assessment Variables")
        fig.update_layout(height=650)
        show_chart(fig)

        st.markdown("### 🔬 Scatter Explorer")
        pretty = {c: c.replace("_", " ").title() for c in rel}
        s1, s2, s3 = st.columns(3)
        xv = s1.selectbox("X axis", rel, index=0, format_func=pretty.get)
        yv = s2.selectbox("Y axis", rel, index=min(5, len(rel) - 1), format_func=pretty.get)
        cv = s3.selectbox("Colour by", ["difficulty_label", "cognitive_level_bloom", "subject"],
                          format_func=lambda v: v.replace("_", " ").title())
        cmap = {"difficulty_label": DIFF_COLORS, "cognitive_level_bloom": BLOOM_COLORS,
                "subject": SUBJ_COLORS}[cv]
        sample = filtered_df if len(filtered_df) <= 3000 else filtered_df.sample(3000, random_state=1)
        fig = px.scatter(sample, x=xv, y=yv, color=cv, color_discrete_map=cmap,
                         opacity=0.6, title=f"{pretty[yv]} vs {pretty[xv]}")
        if xv != yv:
            r = filtered_df[[xv, yv]].corr().iloc[0, 1]
            st.caption(f"Pearson correlation between the two selected variables: r = {r:.3f}. "
                       "Correlation does not prove causation.")
        show_chart(fig)
    else:
        st.info("Not enough varying numeric columns to compute correlations.")

# ---------- TAB 10 ----------
with tabs[9]:
    st.subheader("📋 Question-Level Explorer")
    search_text = st.text_input("🔎 Search question text")
    disp = filtered_df.copy()
    if search_text:
        disp = disp[disp["question_text"].str.contains(search_text, case=False,
                                                       na=False, regex=False)]
    disp = disp[["question_text", "subject", "cognitive_level_bloom", "difficulty_label",
                 "correct_percentage_pct", "learning_gap_pct", "discrimination_index",
                 "quality_pct", "time_taken_minutes"]].copy()
    disp.columns = ["Question", "Subject", "Bloom Level", "Difficulty", "Correct %",
                    "Learning Gap %", "Discrimination", "Quality %", "Time (min)"]
    st.caption(f"{len(disp):,} matching questions")
    show_table(disp.round(3), height=500)
    st.download_button("📥 Download Filtered Questions",
                       data=disp.to_csv(index=False),
                       file_name="filtered_exam_analysis.csv", mime="text/csv")

st.markdown("---")
st.info("🧠 Intelligent Exam Analytics Dashboard • Question-Level Learning Analytics • Built with Streamlit")

'''

APP_PATH = "/content/exam_dashboard.py"
with open(APP_PATH, "w", encoding="utf-8") as f:
    f.write(APP_CODE)

print("\n" + "=" * 60)
print("STREAMLIT APP CREATED:", APP_PATH)
print("=" * 60)

# ------------------------------------------------------------
# 6. STOP OLD PROCESSES
# ------------------------------------------------------------
os.system("pkill -f 'streamlit run' >/dev/null 2>&1 || true")
os.system("pkill -f localtunnel >/dev/null 2>&1 || true")
os.system("pkill -f cloudflared >/dev/null 2>&1 || true")
time.sleep(2)

# ------------------------------------------------------------
# 7. START STREAMLIT (and wait until it is really healthy)
#    CORS / XSRF / websocket-compression are switched off because
#    they break Streamlit when it is opened through a tunnel.
# ------------------------------------------------------------
PORT = 8501
env = os.environ.copy()
env["EXAM_DATA_PATH"] = DATA_PATH

log_path = "/content/streamlit_log.txt"
log_file = open(log_path, "w")
streamlit_process = subprocess.Popen(
    [sys.executable, "-m", "streamlit", "run", APP_PATH,
     "--server.port", str(PORT), "--server.address", "0.0.0.0",
     "--server.headless", "true", "--browser.gatherUsageStats", "false",
     "--server.enableCORS", "false", "--server.enableXsrfProtection", "false",
     "--server.enableWebsocketCompression", "false"],
    stdout=log_file, stderr=subprocess.STDOUT, env=env,
)

print("\nStarting Streamlit...")
healthy = False
for _ in range(60):
    if streamlit_process.poll() is not None:
        break
    try:
        if urllib.request.urlopen(f"http://127.0.0.1:{PORT}/_stcore/health", timeout=2).status == 200:
            healthy = True
            break
    except Exception:
        pass
    time.sleep(1)

if not healthy:
    print("\nStreamlit failed to start.\n===== STREAMLIT LOG =====\n")
    print(open(log_path).read())
    raise RuntimeError("Streamlit did not start correctly.")
print("Streamlit started successfully.")

# ------------------------------------------------------------
# 8. PUBLIC URL  (Cloudflare tunnel first, localtunnel as backup)
# ------------------------------------------------------------
def wait_for_url(log, pattern, timeout=60):
    end = time.time() + timeout
    while time.time() < end:
        try:
            m = re.search(pattern, open(log).read())
            if m:
                return m.group(0)
        except FileNotFoundError:
            pass
        time.sleep(1)
    return None

print("\n" + "=" * 60)
print("STARTING PUBLIC TUNNEL")
print("=" * 60)

public_url, tunnel_name = None, None

# --- option A: Cloudflare quick tunnel (no password, serves all JS files reliably)
cf_bin = "/content/cloudflared"
if not os.path.exists(cf_bin):
    os.system("wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/"
              f"cloudflared-linux-amd64 -O {cf_bin} && chmod +x {cf_bin}")
cf_log = "/content/cf_log.txt"
if os.path.exists(cf_bin):
    subprocess.Popen(f"{cf_bin} tunnel --url http://127.0.0.1:{PORT} "
                     f"--no-autoupdate > {cf_log} 2>&1", shell=True)
    public_url = wait_for_url(cf_log, r"https://[a-z0-9\-]+\.trycloudflare\.com")
    if public_url:
        tunnel_name = "cloudflare"

# --- option B: localtunnel (backup)
if not public_url:
    print("Cloudflare tunnel did not respond - trying localtunnel...")
    os.system("pkill -f cloudflared >/dev/null 2>&1 || true")
    lt_log = "/content/lt_log.txt"
    subprocess.Popen(f"npx --yes localtunnel --port {PORT} > {lt_log} 2>&1", shell=True)
    public_url = wait_for_url(lt_log, r"https://[a-z0-9\-]+\.loca\.lt")
    if public_url:
        tunnel_name = "localtunnel"

# ------------------------------------------------------------
# 9. FINAL RESULT
# ------------------------------------------------------------
print("\n" + "=" * 60)
print("🎉 INTELLIGENT EXAM ANALYTICS DASHBOARD")
print("=" * 60)
if public_url:
    print("\n🌐 PUBLIC DASHBOARD URL:", public_url)
    if tunnel_name == "cloudflare":
        print("   (If the page looks blank for a few seconds, wait and refresh once.)")
    else:
        try:
            ip = urllib.request.urlopen("https://ipv4.icanhazip.com", timeout=10).read().decode().strip()
            print("🔑 If the page asks for a 'Tunnel Password', enter this IP:", ip)
        except Exception:
            print("🔑 If asked for a Tunnel Password, open https://loca.lt/mytunnelpassword")
        print("⚠️ localtunnel can still show 'Failed to fetch dynamically imported module'.")
        print("   If that happens, re-run this cell to get a Cloudflare link instead.")
else:
    print("\n⚠️ Could not create a public URL. Streamlit is running; re-run this cell to retry.")
    try:
        from google.colab.output import eval_js
        print("Colab built-in link:", eval_js(f"google.colab.kernel.proxyPort({PORT})"))
    except Exception:
        pass
```
